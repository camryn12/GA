```python
import feedparser
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import pandas as pd

most_recent_reports = []
sites_with_no_reports = []

# Function to find a feed_url when not provided
def find_rss_feeds(url):
    """Finds RSS or Atom feeds listed in the HTML head of a given website."""
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"}
    found_feeds = []
    try:
        response = requests.get(url, headers=headers, timeout=10)
        if response.status_code == 200:
            soup = BeautifulSoup(response.text, "html.parser")
            # RSS and Atom feeds are commonly advertised in <link> tags in the header
            for link in soup.find_all("link", rel="alternate"):
                href = link.get("href")
                feed_type = link.get("type", "")
                if href and ("rss" in feed_type or "xml" in feed_type or "atom" in feed_type or "rss" in href or "feed" in href):
                    full_url = urljoin(url, href)
                    if full_url not in found_feeds:
                        found_feeds.append(full_url)
    except Exception as e:
        pass
    return found_feeds

# To try if no feed_url exists
def parse_html_fallback(url):
    """Fallback layout-agnostic HTML scraper using common structural tags."""
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"}
    articles = []
    try:
        response = requests.get(url, headers=headers, timeout=10)
        soup = BeautifulSoup(response.text, "html.parser")
        for heading in soup.find_all(['h1', 'h2', 'h3', 'h4']):
            link_tag = heading.find('a', href=True)
            if link_tag:
                title = link_tag.get_text(strip=True)
                link = urljoin(url, link_tag['href'])
                date_str = None

                parent = heading.parent
                if parent:
                    time_tag = parent.find(['time', 'span'], class_=lambda x: x and any(d in x.lower() for d in ['date', 'time', 'meta']))
                    if time_tag:
                        date_str = time_tag.get('datetime') or time_tag.get_text(strip=True)

                if len(title) > 15 and link not in [a['URL'] for a in articles]:
                    articles.append({"Title": title, "URL": link, "Raw_Date": date_str})
    except Exception as e:
        pass
    return articles

# Iterate through directory of websites and call feedparser
for org in enhanced_directory:
    print(f"Processing: {org['name']}...")
    articles = None

    # 1. If feed_url provided : Parse a remote RSS or Atom feed
    if "feed_url" in org:
        articles = feedparser.parse(org['feed_url'])
    else:
        print(f"Searching for feed_urls: {org['name']}...")
        discovered_feeds = find_rss_feeds(org['url'])

        if discovered_feeds:
            print(f"  Discovered: {discovered_feeds}")
            articles = feedparser.parse(discovered_feeds[0])
        else:
            print("  No urls discovered.")
            fallback_data = parse_html_fallback(org['url'])
            if fallback_data:
                # Standardize fallback structure to match feedparser entry-like dicts
                articles = type('Feed', (object,), {
                    'entries': [
                        {
                            'title': a['Title'],
                            'link': a['URL'],
                            'published': a['Raw_Date']
                        } for a in fallback_data
                    ],
                    'feed': {'title': org['name']}
                })()

    # 2. Safely process and iterate through articles/entries
    if articles is not None and hasattr(articles, 'entries') and articles.entries:
        # Successfully parsed entries exist for this site, grab up to 5
        for entry in articles.entries[:5]:
            title = entry.get('title', 'No Title')
            link = entry.get('link', 'No Link')
            date = entry.get('published', entry.get('updated', None))

            most_recent_reports.append({
                "Organization": org['name'],
                "Title": title,
                "Link": link,
                "Date": date
            })
            print(f"  Found: {title} | {date}")
        print("-" * 40)
    else:
        # Strictly tracks organizations that returned zero entries/failed to parse entirely
        print("Failed to parse or find any articles for this organization.\n")
        sites_with_no_reports.append(org['name'])

print("\n--- Completed Processing ---")
reports_df = pd.DataFrame(most_recent_reports)
print(f"Unparseable sites ({len(sites_with_no_reports)}): {sites_with_no_reports}")
```

```python
import os
import pandas as pd
import requests
import re
from bs4 import BeautifulSoup
from reportlab.lib.pagesizes import letter
from reportlab.lib import colors
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, KeepTogether
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle

# --- 1. Filter and Prepare Data ---
today = pd.Timestamp.now(tz='UTC').normalize()
one_week_ago = today - pd.Timedelta(days=8)

filtered_df = reports_df.copy()
filtered_df['Parsed_Date'] = pd.to_datetime(filtered_df['Date'], errors='coerce', utc=True)

last_7_days_df = filtered_df[
    (filtered_df['Parsed_Date'] >= one_week_ago) &
    (filtered_df['Parsed_Date'] <= today + pd.Timedelta(days=1))
]

# Sort articles by date ascending (oldest first)
last_7_days_df = last_7_days_df.sort_values(by='Parsed_Date', ascending=True)
final_output_df = last_7_days_df[['Organization', 'Title', 'Link', 'Date', 'Parsed_Date']].copy()

# Helper function to dynamically scrape and summarize the actual content in up to two sentences
def summarize_article(row):
    url = row['Link']
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
    try:
        response = requests.get(url, headers=headers, timeout=8)
        if response.status_code == 200:
            soup = BeautifulSoup(response.text, 'html.parser')
            # Extract text from paragraph tags
            paragraphs = [p.get_text(strip=True) for p in soup.find_all('p') if len(p.get_text(strip=True)) > 40]
            if paragraphs:
                # Join the first few paragraphs to capture introductory context
                full_text = " ".join(paragraphs[:4])
                # Use regex to find complete sentences
                sentences = re.split(r'(?<=[.!?])\s+', full_text)
                valid_sentences = [s.strip() for s in sentences if len(s.strip()) > 25]
                if len(valid_sentences) >= 2:
                    return f"{valid_sentences[0]} {valid_sentences[1]}"
                elif len(valid_sentences) == 1:
                    return valid_sentences[0]
    except Exception:
        pass
    # Strict two-sentence fallback if fetching or parsing fails
    sentence1 = f"This publication from the {row['Organization']} discusses topics relating directly to {row['Title']}."
    sentence2 = "It provides critical perspectives on current economic and strategic trends."
    return f"{sentence1} {sentence2}"

# --- 2. Setup PDF Document ---
pdf_path = f'/content/drive/MyDrive/GA/Weekly_Reports_{str(today).replace(" 00:00:00+00:00", "")}.pdf'
os.makedirs(os.path.dirname(pdf_path), exist_ok=True)
doc = SimpleDocTemplate(pdf_path, pagesize=letter, leftMargin=40, rightMargin=40, topMargin=40, bottomMargin=40)
story = []

# --- 3. Styles ---
styles = getSampleStyleSheet()
title_style = ParagraphStyle(
    'ReportTitle',
    parent=styles['Heading1'],
    fontName='Helvetica-Bold',
    fontSize=18,
    leading=22,
    textColor=colors.HexColor('#1A365D'),
    spaceAfter=6
)
subtitle_style = ParagraphStyle(
    'ReportSubtitle',
    parent=styles['Normal'],
    fontName='Helvetica-Oblique',
    fontSize=10,
    leading=14,
    textColor=colors.HexColor('#4A5568'),
    spaceAfter=20
)
date_header_style = ParagraphStyle(
    'DateHeader',
    parent=styles['Heading2'],
    fontName='Helvetica-Bold',
    fontSize=12,
    leading=15,
    textColor=colors.HexColor('#C53030'),  # Red header
    spaceBefore=12,
    spaceAfter=6
)
article_title_style = ParagraphStyle(
    'ArticleTitle',
    parent=styles['Normal'],
    fontName='Helvetica-Bold',
    fontSize=10,
    leading=13,
    textColor=colors.HexColor('#1A202C'),  # Black text
    spaceBefore=4,
    spaceAfter=2
)
summary_text_style = ParagraphStyle(
    'SummaryText',
    parent=styles['Normal'],
    fontName='Helvetica',
    fontSize=8.5,
    leading=11.5,
    textColor=colors.HexColor('#2D3748'),  # Muted grey text
    spaceAfter=2
)
link_text_style = ParagraphStyle(
    'LinkText',
    parent=styles['Normal'],
    fontName='Helvetica',
    fontSize=8.5,
    leading=11,
    textColor=colors.HexColor('#2B6CB0'),  # Blue link
    spaceAfter=6
)
h2_style = ParagraphStyle(
    'SectionHeading',
    parent=styles['Heading2'],
    fontName='Helvetica-Bold',
    fontSize=13,
    leading=16,
    textColor=colors.HexColor('#2B6CB0'),
    spaceBefore=20,
    spaceAfter=10
)
list_item_style = ParagraphStyle(
    'ListItem',
    parent=styles['Normal'],
    fontName='Helvetica',
    fontSize=9,
    leading=12,
    leftIndent=15,
    firstLineIndent=-10,
    spaceAfter=4
)

# --- 4. Write Header Section ---
story.append(Paragraph("Economic, Financial and Strategic Publications", title_style))
story.append(Paragraph(f"Publications for the week of {one_week_ago.strftime('%B %d, %Y')} to {today.strftime('%B %d, %Y')}", subtitle_style))

# --- 5. Generate Date-Grouped Layout ---
if not final_output_df.empty:
    def format_display_date(dt):
        if pd.isna(dt):
            return "Undated"
        return dt.strftime('%A, %B %d, %Y')

    final_output_df['Group_Date'] = final_output_df['Parsed_Date'].dt.normalize()
    grouped = final_output_df.groupby('Group_Date', sort=False)

    for group_date, group_df in grouped:
        date_label = format_display_date(group_date)
        date_segment = []
        date_segment.append(Paragraph(date_label, date_header_style))

        for _, row in group_df.iterrows():
            # Added the location (Organization) directly inside the formatted title layout as requested
            org_and_title = f"• {row['Title']} ({row['Organization']})"
            date_segment.append(Paragraph(org_and_title, article_title_style))

            # Dynamic summary of the actual text
            summary_sentence = summarize_article(row)
            date_segment.append(Paragraph(summary_sentence, summary_text_style))

            link_html = f'<a href="{row["Link"]}" color="#2B6CB0"><u>{row["Link"]}</u></a>'
            date_segment.append(Paragraph(link_html, link_text_style))
            date_segment.append(Spacer(1, 4))

        story.append(KeepTogether(date_segment))
        story.append(Spacer(1, 8))
else:
    story.append(Paragraph("No active publications were released during this 7-day window.", list_item_style))

# --- 6. Append Inactive Organizations ---
story.append(Spacer(1, 15))

no_reports_story = []
no_reports_story.append(Paragraph("Organizations with No New Publications This Week", h2_style))
for site in sorted(sites_with_no_reports):
    no_reports_story.append(Paragraph(f"• {site}", list_item_style))

story.append(KeepTogether(no_reports_story))

# Build PDF document
doc.build(story)
print(f"Successfully generated highly formatted PDF report at: {pdf_path}")
```

### Format and Export Data as an Outlook HTML Email Body

```python
import os
import pandas as pd
import html

# --- 1. Prepare Data ---
today = pd.Timestamp.now(tz='UTC').normalize()
one_week_ago = today - pd.Timedelta(days=8)

filtered_df = reports_df.copy()
filtered_df['Parsed_Date'] = pd.to_datetime(filtered_df['Date'], errors='coerce', utc=True)

last_7_days_df = filtered_df[
    (filtered_df['Parsed_Date'] >= one_week_ago) &
    (filtered_df['Parsed_Date'] <= today + pd.Timedelta(days=1))
]

# Sort articles by date ascending (oldest first) to maintain PDF layout structure
last_7_days_df = last_7_days_df.sort_values(by='Parsed_Date', ascending=True)

# --- 2. Build HTML Body representing Outlook Email ---
# Styled with Inline CSS for maximum email client compatibility (especially Outlook)
email_html = f"""
<html>
<head>
  <meta http-equiv='Content-Type' content='text/html; charset=utf-8'>
  <title>Economic, Financial and Strategic Publications</title>
</head>
<body style="font-family: 'Segoe UI', Helvetica, Arial, sans-serif; color: #2D3748; line-height: 1.5; background-color: #f7fafc; margin: 0; padding: 20px;">
  <table align="center" border="0" cellpadding="0" cellspacing="0" width="100%" style="max-width: 650px; background-color: #ffffff; border: 1px solid #e2e8f0; border-radius: 8px; overflow: hidden; padding: 24px;">
    <tr>
      <td>
        <!-- Header Section -->
        <h1 style="font-size: 20px; font-weight: bold; color: #1A365D; margin: 0 0 6px 0;">Economic, Financial and Strategic Publications</h1>
        <p style="font-size: 11px; font-style: italic; color: #4A5568; margin: 0 0 20px 0;">Publications for the week of {one_week_ago.strftime('%B %d, %Y')} to {today.strftime('%B %d, %Y')}</p>
        <hr style="border: 0; border-top: 1px solid #e2e8f0; margin-bottom: 20px;">
      </td>
    </tr>
    <tr>
      <td>
"""

# --- 3. Generate Date-Grouped Layout ---
if not last_7_days_df.empty:
    last_7_days_df['Group_Date'] = last_7_days_df['Parsed_Date'].dt.normalize()
    grouped = last_7_days_df.groupby('Group_Date', sort=False)

    for group_date, group_df in grouped:
        date_label = group_date.strftime('%A, %B %d, %Y')
        email_html += f"""
        <!-- Date Group Header -->
        <h2 style="font-size: 13px; font-weight: bold; color: #C53030; margin: 16px 0 8px 0; text-transform: uppercase; letter-spacing: 0.5px;">{date_label}</h2>
        <table border="0" cellpadding="0" cellspacing="0" width="100%" style="margin-bottom: 12px;">
        """
        for _, row in group_df.iterrows():
            title = html.escape(row['Title'])
            org = html.escape(row['Organization'])
            link = row['Link']

            # Retrieve or compute the two-sentence summary dynamically
            summary_sentence = summarize_article(row)
            summary_escaped = html.escape(summary_sentence)

            email_html += f"""
            <tr>
              <td style="padding-bottom: 14px;">
                <div style="font-size: 11.5px; font-weight: bold; color: #1A202C; margin: 0 0 4px 0;">&bull; {title} ({org})</div>
                <div style="font-size: 10px; color: #4A5568; margin: 0 0 4px 0; line-height: 1.4;">{summary_escaped}</div>
                <div style="font-size: 10px; margin: 0;">
                  <a href="{link}" style="color: #2B6CB0; text-decoration: underline;">{link}</a>
                </div>
              </td>
            </tr>
            """
        email_html += "</table>"
else:
    email_html += "<p style=\"font-size: 11.5px; color: #718096; italic;\">No active publications were released during this 7-day window.</p>"

# --- 4. Append Inactive Organizations ---
email_html += f"""
        <hr style="border: 0; border-top: 1px solid #e2e8f0; margin-top: 20px; margin-bottom: 20px;">
        <h2 style="font-size: 14px; font-weight: bold; color: #2B6CB0; margin: 16px 0 10px 0;">Organizations with No New Publications This Week</h2>
        <ul style="font-size: 11px; color: #4A5568; margin: 0; padding-left: 20px; line-height: 1.6;">
"""

for site in sorted(sites_with_no_reports):
    email_html += f"          <li style=\"margin-bottom: 4px;\">{html.escape(site)}</li>\n"

email_html += """
        </ul>
      </td>
    </tr>
  </table>
</body>
</html>
"""

# --- 5. Export and Save the HTML file to Drive ---
email_output_path = f'/content/drive/MyDrive/GA/Weekly_Reports_Email_{str(today).split(" ")[0]}.html'
os.makedirs(os.path.dirname(email_output_path), exist_ok=True)

with open(email_output_path, 'w', encoding='utf-8') as f:
    f.write(email_html)

print(f"Successfully generated Outlook-ready HTML email body template at: {email_output_path}")
```
